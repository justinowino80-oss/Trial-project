```python
**For Loops**
A loop is a way of telling Python to repeat something.
The two main types of loops are:
- for loop
- while loop

    1. FOR LOOP
A for loop is normally used when you want to go through a collection of items, such as a list, one item at a time.

Simple example
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)

Output:
apple
banana
cherry

Python goes through the list like this:
First time → fruit = "apple"
Second time → fruit = "banana"
Third time → fruit = "cherry"
```

```python
FOR loop with range()
You will also see for loops used with range().
    
    For example:
for number in range(5):
    print(number)
Output:
0
1
2
3
4

If you want 1 to 5:
for number in range(1, 6):
    print(number)
Output:
1
2
3
4
5
```

```python
2. WHILE LOOP
A while loop repeats something as long as a condition is true.
    
    For example:
number = 1
while number <= 5:
    print(number)
    number = number + 1

    Output:
1
2
3
4
5

Let's break this down carefully.

- First line
number = 1
We start with number equal to 1.

- Second line
while number <= 5:
This means:
"Keep repeating while number is less than or equal to 5."

-Third line
print(number)
Print the current number.

- Fourth line
number = number + 1
Increase the number by 1.

So Python does:
number = 1 → print 1
number = 2 → print 2
number = 3 → print 3
number = 4 → print 4
number = 5 → print 5

When number becomes 6:
6 <= 5
is False, so Python stops the loop.
```

### Grouping and Aggregation

Breaks down your data into separate groups based on matching values.
Aggregation is to work/calculate on those indivial gruops that your data is broken(bunch together and get summary number, like total, average)

```python
# average tip for each of the days
# grouping by single column
df3.groupby("day")["tip"].mean()
```

`df3.groupby("day")["total_bill"].sum():`This code calculates the total bill amount for each day in a dataset.

`.groupby("day")`
This groups all rows that have the same day together.
For example, all Sunday records are grouped together, as are all Monday records.

`["total_bill"]`
This selects the total_bill column because that is the column we want to calculate.

`.sum()`
This adds together the bill amounts within each group.

Think of it as:
`Group → Select → Add`
- `groupby()` — group similar records.
- `["amount"]` — select the column to calculate.
- `sum()` — add the values.
You can use the same pattern for sales by product, expenses by department, or survey records by village.

**How to rename output columns**

Renaming output columns based on this code
df3.groupby('time').agg(
    min_tip=('tip', 'min'),
    avg_bill=('total_bill', 'mean')
)

The main idea is that .agg() allows you to perform calculations and give the resulting columns your own names.

1. Understand the code step by step

Step 1: `df3.groupby('time')`
This groups the data according to the time column.
For example, the records might be grouped into:
- Lunch
- Dinner
  
Step 2: `.agg()`
agg is short for aggregate. It allows you to perform calculations on groups of data.
You can calculate the minimum, maximum, average, sum, and other statistics.

Step 3: `min_tip=('tip', 'min')`

Let's break this into three parts:
Code	Meaning

`min_tip	The new name of the output column`
`'tip'	The original column to use`
`'min'	The calculation to perform`
In simple language:
Take the tip column, find the minimum tip for each time group, and name the result `min_tip.`

Step 4: `avg_bill=('total_bill', 'mean')`
Code	Meaning

`avg_bill`	The new name of the output column
`'total_bill'`	The original column to use
'mean'	Calculate the average
In simple language:
Take the total_bill column, calculate the average bill for each time group, and name the result avg_bill.

2. See an example
Suppose your data looks like this:
time	tip	total_bill
Lunch	2	20
Lunch	4	30
Dinner	3	40
Dinner	6	50

When you run your code, the output will be:

time	min_tip	avg_bill
Dinner	3	45
Lunch	2	25
`Notice that min_tip and avg_bill` are names you chose. They do not have to match the original column names.

4. The important pattern to remember
`new_name=('original_column', 'calculation')`



For example:
max_tip=('tip', 'max')total_sales=('total_bill', 'sum')avg_tip=('tip', 'mean')


These would create output columns named max_tip, total_sales, and avg_tip.
Remember: `The name on the left` becomes the `new output column name`. The first item inside the parentheses identifies `the original column`, and the second `specifies the calculation.`

```python
df3.groupby('time').agg(
    min_tip=('tip', 'min'),
    avg_bill=('total_bill', 'mean')
)
```

**Sub plots**
This is where you can have many graphs in 1 image

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [100, 120, 90, 150, 180]

regions = ['North', 'South', 'East', 'West']
profits = [12450, -2100, 6500, 8900]

sales_scatter = [100, 250, 75, 400, 120, 300, 80, 500]
profit_scatter = [10, 50, -5, 80, 20, 60, -2, 95]

fixed_sales_data = [60, 85, 90, 95, 100, 105, 110, 115, 120, 130, 140, 150, 160, 180, 210]
mean_value = sum(fixed_sales_data) / len(fixed_sales_data)

fig,axes=plt.subplots(2,2,figsize=(12,10)) # this generates the number of plots you want
# plot 1: Line chart(top left)
axes[0,0].plot(months,sales,color="orange")
axes[0,0].set_title("Sales Trend")
axes[0,0].set_ylabel("Sales")
axes[0,0].set_xlabel("Months")
axes[0,0].grid(alpha=0.4)

# Plot 2 bar graph(top right)
axes[0,1].bar(regions,profits)
axes[0,0].set_title("Profits per Region")
axes[0,0].set_ylabel("Profits")
axes[0,0].set_xlabel("Regions")

# Scatter plots(bottom left)
axes[1,0].scatter(sales_scatter,profit_scatter)
axes[1,0].set_title("Sales vs Profits")
axes[1,0].set_ylabel("Profits")
axes[1,0].set_xlabel("Sales")

# Histogram(bottom right)
axes[1,1].hist(fixed_sales_data,bins=10,color="brown")
axes[1,1].set_title("Fixed Sales")
axes[1,1].set_ylabel("Number")
axes[1,1].set_xlabel("Frequency")
axes[1,1].axvline(mean_value,linestyle="--",color="yellow")

plt.savefig("Combined_graphs.png",dpi=300)

plt.show()
```
